```python
%sql
CREATE OR REPLACE TABLE dbjahan.Silver_data.folkmangd AS
SELECT
  Region                         AS kommun_kod,
  CAST(Tid AS INT)               AS ar,
  SUM(Folkmangd)                 AS folkmangd
FROM dbjahan.bronze_data.folkmangd_raw
WHERE length(Region) = 4
  AND Alder = 'tot'
GROUP BY Region, Tid;
```

```python
%sql
SELECT * FROM dbjahan.Silver_data.folkmangd
LIMIT 10;
```
