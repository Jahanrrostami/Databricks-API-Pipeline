```python
%sql
CREATE OR REPLACE TABLE dbjahan.silver_data.elpris AS
WITH timmar AS (
  SELECT
    elomrade,
    substr(tid_start, 1, 13) AS timme,
    AVG(sek_per_kwh) AS sek_per_kwh
  FROM dbjahan.bronze_data.elpris_raw
  WHERE sek_per_kwh IS NOT NULL
    AND substr(tid_start, 1, 4) = '2025'
  GROUP BY elomrade, substr(tid_start, 1, 13)
)
SELECT
  elomrade,
  2025 AS ar,
  ROUND(AVG(sek_per_kwh), 3) AS snitt_sek_per_kwh,
  COUNT(*) AS n_timmar
FROM timmar
GROUP BY elomrade;
```

```python
%sql
SELECT * FROM dbjahan.Silver_data.elpris
LIMIT 10;
```
