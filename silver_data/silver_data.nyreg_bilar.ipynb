```python
%sql
CREATE OR REPLACE TABLE dbjahan.Silver_data.nyreg_bilar AS
SELECT
  region                         AS kommun_kod,
  region_text                    AS kommun_namn,
  drivmedel                      AS drivmedel_kod,
  drivmedel_text                 AS drivmedel_namn,
  CAST(left(tid, 4) AS INT)      AS ar,
  CAST(right(tid, 2) AS INT)     AS manad,
  tid,
  antal
FROM dbjahan.bronze_data.nyreg_bilar_raw
WHERE length(region) = 4
  AND antal IS NOT NULL;
```

```python
%sql
SELECT * FROM dbjahan.Silver_data.nyreg_bilar
LIMIT 10;
```
