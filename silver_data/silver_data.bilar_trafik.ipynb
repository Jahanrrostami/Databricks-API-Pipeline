```python
%sql
CREATE OR REPLACE TABLE dbjahan.Silver_data.bilar_trafik AS
SELECT
  region                         AS kommun_kod,
  region_text                    AS kommun_namn,
  CAST(tid AS INT)               AS ar,
  antal                          AS bilar_i_trafik
FROM dbjahan.bronze_data.bilar_trafik_raw
WHERE length(region) = 4
  AND agarkategori = '000';
```

```python
%sql
SELECT * FROM dbjahan.Silver_data.bilar_trafik
LIMIT 10;
```
